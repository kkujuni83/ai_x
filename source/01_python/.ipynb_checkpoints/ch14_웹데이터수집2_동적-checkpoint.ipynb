페이징 처리
- 다음 뉴스 검색 : 원하는 단어를 원하는 page수만큼

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
driver = webdriver.Chrome()
driver.get('https://www.daum.net/')
# input 태그를 click
time.sleep(1)
driver.find_element(By.NAME, 'q').click()
query = input('검색하고자 하는 단어?')
driver.find_element(By.CSS_SELECTOR, 'input[type="text"]').send_keys(query)
driver.find_element(By.CLASS_NAME, 'btn_ksearch').click()
time.sleep(2)
# 뉴스탭 클릭 : ul.list_tab > li
# driver.find_elements(By.CSS_SELECTOR, 'ul.list_tab > li')[3].click()
driver.find_element(By.LINK_TEXT, '뉴스').click() # 쉬는 시간 훈련생 제안
time.sleep(2)
news_list = []
pages = 3
for page in range(1, pages+1):
    elems = driver.find_elements(By.CSS_SELECTOR, 'div.item-title > strong.tit-g > a')
    for elem in elems:
        title = elem.text
        link  = elem.get_attribute('href')
        news_list.append([title, link])
        # print(title, link)
    # 2page로
    page_div = driver.find_element(By.CSS_SELECTOR, 'div.inner_paging')
    # print(page_div.text)
    next_page = page_div.find_element(By.LINK_TEXT, str(page+1) )
    next_page.click()
    time.sleep(1.5)
    print(f'~~ 현재 {page}페이지 데이터 수집 중입니다 ~ ~')
driver.close()
import pandas as pd
df = pd.DataFrame(news_list, columns=['title', 'link'])
df.to_csv('data/ch14_daum.csv', index=False)
```

# 4절-2 연습문제
영어번역 자동화 프로그램을 구현하시오. 네이버 맞춤법 검사를 마친 텍스트를 https://translate.kakao.com/를 통해 영어로 번역된 텍스트    
를 파일로 출력하는 동적 웹크롤링 프로그램을 구현하시오.

- 번역전 파일 : ch14_맞춤법후.txt
- 번역후 파일 : 2_자동화영어번역본.txt

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import os

def translate_text_with_kakao():
    # Chrome 옵션 설정
    chrome_options = Options()
    chrome_options.add_argument("--no-sandbox")
    chrome_options.add_argument("--disable-dev-shm-usage")
    chrome_options.add_argument("--disable-blink-features=AutomationControlled")
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    
    driver = webdriver.Chrome(options=chrome_options)
    driver.execute_script("Object.defineProperty(navigator, 'webdriver', {get: () => undefined})")
    
    try:
        # 입력 파일 읽기
        input_file = r'C:\ai_x\source\01_python\data\ch14_맞춤법후.txt'
        if not os.path.exists(input_file):
            print(f"입력 파일 '{input_file}'을 찾을 수 없습니다.")
            return
        
        with open(input_file, 'r', encoding='utf-8') as f:
            korean_text = f.read().strip()
        
        if not korean_text:
            print("번역할 텍스트가 없습니다.")
            return
        
        print("카카오 번역 사이트 접속 중...")
        driver.get('https://translate.kakao.com/')
        
        # 페이지 로딩 대기
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "body"))
        )
        time.sleep(3)
        
        # 텍스트를 1000자 이하로 분할 (띄어쓰기 포함)
        max_chars = 1000
        text_chunks = []
        
        print(f"원본 텍스트 길이: {len(korean_text)}자")
        
        if len(korean_text) <= max_chars:
            text_chunks = [korean_text]
        else:
            # 문장 단위로 분할하여 1000자 이하 청크 생성
            sentences = korean_text.replace('.\n', '.|SPLIT|').replace('. ', '.|SPLIT|').replace('!\n', '!|SPLIT|').replace('? ', '?|SPLIT|').split('|SPLIT|')
            current_chunk = ""
            
            for sentence in sentences:
                sentence = sentence.strip()
                if not sentence:
                    continue
                    
                # 현재 청크에 문장을 추가했을 때 1000자를 초과하는지 확인
                test_chunk = current_chunk + (' ' if current_chunk else '') + sentence
                
                if len(test_chunk) > max_chars:
                    # 현재 청크가 비어있지 않으면 저장
                    if current_chunk.strip():
                        text_chunks.append(current_chunk.strip())
                        current_chunk = sentence
                    else:
                        # 단일 문장이 1000자를 초과하는 경우 강제 분할
                        while len(sentence) > max_chars:
                            # 적절한 위치에서 분할 (공백 기준)
                            split_pos = sentence.rfind(' ', 0, max_chars)
                            if split_pos == -1:
                                split_pos = max_chars
                            
                            text_chunks.append(sentence[:split_pos].strip())
                            sentence = sentence[split_pos:].strip()
                        
                        current_chunk = sentence
                else:
                    current_chunk = test_chunk
            
            # 마지막 청크 추가
            if current_chunk.strip():
                text_chunks.append(current_chunk.strip())
        
        # 각 청크의 길이 확인
        for i, chunk in enumerate(text_chunks, 1):
            print(f"청크 {i}: {len(chunk)}자")
            if len(chunk) > max_chars:
                print(f"  경고: 청크 {i}이 {max_chars}자를 초과합니다!")
        
        print(f"총 {len(text_chunks)}개의 텍스트 블록으로 분할하여 번역합니다. (각 블록 최대 1000자)")
        
        translated_results = []
        
        for i, chunk in enumerate(text_chunks, 1):
            print(f"번역 중... ({i}/{len(text_chunks)})")
            
            try:
                # 입력 텍스트 영역 찾기 (여러 선택자 시도)
                input_selectors = [
                    'textarea[placeholder*="번역할"]',
                    'textarea.txt_source',
                    '#translateTextarea',
                    'textarea[name="source"]',
                    '.input_translate textarea'
                ]
                
                input_textarea = None
                for selector in input_selectors:
                    try:
                        input_textarea = WebDriverWait(driver, 5).until(
                            EC.element_to_be_clickable((By.CSS_SELECTOR, selector))
                        )
                        print(f"입력 영역 찾음: {selector}")
                        break
                    except:
                        continue
                
                if not input_textarea:
                    print("입력 텍스트 영역을 찾을 수 없습니다.")
                    # 페이지 소스 확인을 위한 디버깅
                    soup = BeautifulSoup(driver.page_source, "html.parser")
                    textareas = soup.find_all('textarea')
                    print(f"페이지에서 찾은 textarea 개수: {len(textareas)}")
                    for idx, ta in enumerate(textareas):
                        print(f"Textarea {idx}: {ta.get('class', 'no-class')} - {ta.get('placeholder', 'no-placeholder')[:50]}")
                    
                    # 첫 번째 textarea 사용 시도
                    if textareas:
                        try:
                            input_textarea = driver.find_element(By.TAG_NAME, 'textarea')
                        except:
                            pass
                
                if input_textarea:
                    # 기존 텍스트 클리어
                    input_textarea.click()
                    time.sleep(0.5)
                    input_textarea.send_keys(Keys.CONTROL, 'a')
                    time.sleep(0.3)
                    input_textarea.send_keys(Keys.DELETE)
                    time.sleep(0.5)
                    
                    # 새 텍스트 입력
                    input_textarea.send_keys(chunk)
                    time.sleep(2)
                    
                    # 번역 버튼 클릭 시도
                    button_selectors = [
                        'button[class*="translate"]',
                        '.btn_translate',
                        'button[type="submit"]',
                        '.translate_btn'
                    ]
                    
                    button_clicked = False
                    for selector in button_selectors:
                        try:
                            button = driver.find_element(By.CSS_SELECTOR, selector)
                            if button.is_enabled():
                                button.click()
                                button_clicked = True
                                print("번역 버튼 클릭됨")
                                break
                        except:
                            continue
                    
                    # Enter 키로 번역 시도
                    if not button_clicked:
                        input_textarea.send_keys(Keys.RETURN)
                        print("Enter 키로 번역 시도")
                    
                    # 번역 결과 대기
                    time.sleep(4)
                    
                    # 번역 결과 가져오기
                    result_selectors = [
                        'textarea[readonly]',
                        '.txt_target',
                        'textarea.txt_target',
                        '#result_translate',
                        '.result_translate textarea'
                    ]
                    
                    translated_text = ""
                    for selector in result_selectors:
                        try:
                            result_element = driver.find_element(By.CSS_SELECTOR, selector)
                            translated_text = result_element.get_attribute('value') or result_element.text
                            if translated_text and translated_text.strip() and translated_text != chunk:
                                print(f"번역 결과 찾음: {selector}")
                                break
                        except:
                            continue
                    
                    # BeautifulSoup으로 재시도
                    if not translated_text or translated_text.strip() == chunk:
                        soup = BeautifulSoup(driver.page_source, "html.parser")
                        
                        # 모든 textarea 확인
                        textareas = soup.find_all('textarea')
                        for ta in textareas:
                            text_content = ta.get_text().strip()
                            if text_content and text_content != chunk and len(text_content) > 10:
                                translated_text = text_content
                                print("BeautifulSoup으로 번역 결과 찾음")
                                break
                    
                    if translated_text and translated_text.strip() and translated_text != chunk:
                        translated_results.append(translated_text.strip())
                        print(f"블록 {i} 번역 완료 (입력: {len(chunk)}자 → 출력: {len(translated_text.strip())}자)")
                    else:
                        print(f"블록 {i} 번역 결과를 가져올 수 없습니다.")
                        print(f"입력 길이: {len(chunk)}자")
                        print(f"찾은 텍스트: {translated_text[:100] if translated_text else '없음'}")
                        translated_results.append(f"[번역 실패: 블록 {i} - {len(chunk)}자]\n{chunk}")
                
                else:
                    print("입력 영역을 찾을 수 없어 해당 블록을 건너뜁니다.")
                    translated_results.append(f"[번역 실패: 블록 {i} - {len(chunk)}자]\n{chunk}")
                
            except Exception as e:
                print(f"블록 {i} 번역 중 오류 발생: {e}")
                translated_results.append(f"[번역 실패: 블록 {i} - {len(chunk)}자]\n{chunk}")
            
            time.sleep(2)  # 요청 간격 조절
        
        # 결과 파일 저장 (52자 내외로 줄바꿈 적용)
        output_dir = r'C:\ai_x\source\01_python\data'
        if not os.path.exists(output_dir):
            os.makedirs(output_dir)
        
        output_file = os.path.join(output_dir, '2_자동화영어번역본.txt')
        
        # 번역 결과를 95자 내외로 줄바꿈 처리
        def format_text_with_line_breaks(text, max_line_length=95):
            """텍스트를 지정된 길이로 줄바꿈하되 단어 단위로 자름"""
            paragraphs = text.split('\n\n')
            formatted_paragraphs = []
            
            for paragraph in paragraphs:
                if not paragraph.strip():
                    continue
                    
                # 기존 줄바꿈 제거하고 한 줄로 만들기
                clean_text = ' '.join(paragraph.split())
                
                lines = []
                current_line = ""
                words = clean_text.split(' ')
                
                for word in words:
                    # 현재 줄에 단어를 추가했을 때의 길이 계산
                    test_line = current_line + (' ' if current_line else '') + word
                    
                    if len(test_line) <= max_line_length:
                        current_line = test_line
                    else:
                        # 현재 줄이 비어있지 않으면 저장
                        if current_line:
                            lines.append(current_line)
                            current_line = word
                        else:
                            # 단어가 너무 긴 경우 그대로 추가
                            lines.append(word)
                            current_line = ""
                
                # 마지막 줄 추가
                if current_line:
                    lines.append(current_line)
                
                formatted_paragraphs.append('\n'.join(lines))
            
            return '\n\n'.join(formatted_paragraphs)
        
        # 모든 번역 결과를 합치고 포맷팅
        final_result = '\n\n'.join(translated_results)
        formatted_result = format_text_with_line_breaks(final_result)
        
        with open(output_file, 'w', encoding='utf-8') as f:
            f.write(formatted_result)
        
        print(f"\n번역 완료! 결과가 '{output_file}'에 저장되었습니다.")
        print(f"총 번역된 블록 수: {len(translated_results)}")
        print("번역된 텍스트는 95자 내외로 줄바꿈하여 가독성을 높였습니다.")
        
    except Exception as e:
        print(f"프로그램 실행 중 오류 발생: {e}")
        print("페이지 소스 일부 확인:")
        try:
            soup = BeautifulSoup(driver.page_source, "html.parser")
            print(f"페이지 제목: {soup.title.string if soup.title else '제목 없음'}")
            textareas = soup.find_all('textarea')
            print(f"Textarea 개수: {len(textareas)}")
        except:
            print("페이지 소스 확인 실패")
    
    finally:
        driver.quit()
        print("브라우저 종료")

# 실행 부분
if __name__ == "__main__":
    translate_text_with_kakao()
```
